# 주피터 노트북 시작하기

```python
import os           # 운영체제 관련 표준 라이브러리
import pathlib      # 파일 경로를 객체로 다루는 표준 라이브러리

here = pathlib.Path.cwd()   # 현재 작업 위치 확인

ROOT = here.parents[2] if here.name == "day01" else here    # hanwha-agent

os.chdir(ROOT)      # 앞으로 모든 상대경로는 이 폴더가 기준이된다.

SANDBOX = ROOT / "sandbox" / "w2" / "day01"
# SANDBOX.mkdir(parents=True, exist_ok=True)
```

```python
print(ROOT)
print(SANDBOX.relative_to(ROOT))
```

```python
(SANDBOX / "hello_mod.py").write_text("MESSAGE = '나는 sandbox/w2/day01 안에 있다.'", encoding="utf-8")
print("만든파일 : ", (SANDBOX / "hello_mod.py").relative_to(ROOT))
```
