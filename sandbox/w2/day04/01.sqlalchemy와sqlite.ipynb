```python
import os      
import pathlib   

here = pathlib.Path.cwd()  # 현재 작업 위치 확인

ROOT = here.parents[2] if here.name == "day04" else here # hanwha-agent 

os.chdir(ROOT) # 루트 디렉토리 설정 : 앞으로 모든 상대경로는 이 폴더가 기준이된다.
```

```python
SANDBOX = ROOT / "sandbox" / "w2" / "day04"

# SQLite : 파이썬 표준 라이브러리. 설치 필요없음. 
import sqlite3

# 실습용 DB 파일 생성 경로 
db_path = SANDBOX / "sql_practice.db"

# db 파일 연결 : 파일이 없으면 새로 만들어짐 
conn = sqlite3.connect(db_path)

# 커서 cursor : DB 에 명령을 보내고 결과를 받아오는 창구 
cur = conn.cursor() 

print("연결 완료 : ", db_path)
```

```python
sql = """
    CREATE TABLE IF NOT EXISTS documents(
    id INTEGER PRIMARY KEY, 
    doc_id VARCHAR(20) NOT NULL, 
    title VARCHAR(200) NOT NULL, 
    version VARCHAR(10) NOT NULL, 
    department VARCHAR(15) NOT NULL, 
    security_level VARCHAR(10) NOT NULL, 
    effective_date DATE NOT NULL, 
    expiry_date DATE, 
    is_latest BOOLEAN NOT NULL DEFAULT FALSE, 
    page_count INTEGER NOT NULL DEFAULT 0, 
    create_at TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP 
);
"""
cur.execute(sql)

print("DB 테이블 생성")

# 어떤 테이블이 있는지 확인
cur.execute("SELECT name FROM sqlite_master WHERE type='table';")
print("테이블 목록 : ", cur.fetchall())
```

```python
# 데이터 저장
sql = """
INSERT INTO documents(doc_id, title, version, department, security_level, effective_date, expiry_date, is_latest, page_count) 
VALUES(?,?,?,?,?,?,?,?,?); 
""" 
cur.execute(sql, ("DOC-HR-012", "출장 여비 규정", "1.0", "인사팀", "일반", "2025-01-01", "2025-12-31", False, 28))
print("저장 완료")
conn.commit()
```

```python
# 데이터 조회 
sql = "SELECT id, doc_id, title, version, department FROM documents;"
cur.execute(sql, ())

result = cur.fetchall() 
print(result)

for r in result:
    id = r[0]
    doc_id = r[1]
    title = r[2]
    print(id, doc_id, title)

conn.close() 
```

```python
# SQL Alchemy 조회 
from sqlalchemy import create_engine, text
from sqlalchemy.orm import sessionmaker

engine = create_engine("sqlite:///C:/hw/hw/python_basic/hanhwa-agent/sandbox/w2/day04/sql_practice.db")
SessionLocal = sessionmaker(bind=engine)

with SessionLocal() as session:
    rows = session.execute(
        text("SELECT id, doc_id, title, version, department FROM documents")
    ).all()

    session.commit()
```
